```python
import tensorflow as tf
from sklearn.datasets import fetch_openml
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
Dataset = fetch_openml('mnist_784',as_frame=False)

X = Dataset.data.astype("float32") / 255.0
y = Dataset.target.astype("int32")

x_train_UnsupervisedLearning,x_train_SupervisedLearning,y_train_UnsupervisedLearning,y_train_SupervisedLearning = train_test_split(X,y,random_state=42,stratify=y,test_size=0.01)
x_train_UnsupervisedLearning, x_test, y_train_UnsupervisedLearning, y_test = train_test_split(x_train_UnsupervisedLearning, y_train_UnsupervisedLearning,
    random_state=42,stratify=y_train_UnsupervisedLearning, test_size=0.1)

print('shape x_train_SupervisedLearning is : ',x_train_SupervisedLearning.shape)
print('shape x_train_UnsupervisedLearning is : ',x_train_UnsupervisedLearning.shape)
print('shape x_test is : ', x_test.shape)
print('Total sum train_test_split of x : ',x_test.shape[0]+x_train_UnsupervisedLearning.shape[0]+x_train_SupervisedLearning.shape[0])

tf.keras.utils.set_random_seed(42)
```

```python
i =0 
class_1_TO_9  = 0
list_class_0_TO_9 =[]

while len( list_class_0_TO_9 ) < 10 :
    for classs in y_train_UnsupervisedLearning:
        if classs == class_1_TO_9 :
            list_class_0_TO_9 .append(i)
            class_1_TO_9 +=1
        i+=1


fig, axs = plt.subplots(1,10,figsize=(15,3))

for i , c in enumerate(list_class_0_TO_9):
    axs[i].imshow(x_train_UnsupervisedLearning[c].reshape(28, 28), cmap="Blues")
    axs[i].axis('off')
plt.show()
```

```python
class FNN(tf.keras.Model):

    def __init__(
            self,
            n_Nrural_output=10,
            n_Neural=128,
            activation='relu',
            Number=tf.keras.initializers.HeNormal(),
            n_layers=3,
            shape_input=(28,28),
            activation_output='softmax',
            **kwargs,
    ):
        
        super().__init__(**kwargs)

        self.n_Nrural_output = n_Nrural_output
        self.n_Neural = n_Neural
        self.activation = activation
        self.Number = Number
        self.n_layers = n_layers
        self.shape_input = shape_input
        self.activation_output = activation_output
        self.flatten = tf.keras.layers.Flatten()

        self.hidden_layers = []

        for i in range(self.n_layers):
            self.hidden_layers.append(
                tf.keras.layers.Dense(
                    self.n_Neural,
                    activation=self.activation,
                    kernel_initializer=self.Number
                )
            )

        self.output_layer = tf.keras.layers.Dense(
            self.n_Nrural_output,
            activation=self.activation_output
        )

    def call(self, inputs):

        x = self.flatten(inputs)

        for layer in self.hidden_layers:
            x = layer(x)

        return self.output_layer(x)


model_classFNN = FNN(
    n_Neural=100,
    activation='relu',
    Number=tf.keras.initializers.HeNormal(),
    n_layers=3,
    activation_output='softmax',
    shape_input=(28,28),
    n_Nrural_output=10
)

model_classFNN.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
model_classFNN.fit(x_train_SupervisedLearning,y_train_SupervisedLearning,epochs=10)

test_loss, test_accuracy = model_classFNN.evaluate(x_test,y_test,verbose=0)

print("Test Accuracy Class_FNN:", test_accuracy)
```

```python
model_FNN = tf.keras.models.Sequential([
    tf.keras.Input(shape=(784,)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(100,activation='relu',kernel_initializer=tf.keras.initializers.HeNormal()),
    tf.keras.layers.Dense(100,activation='relu',kernel_initializer=tf.keras.initializers.HeNormal()),
    tf.keras.layers.Dense(100,activation='relu',kernel_initializer=tf.keras.initializers.HeNormal()),
    tf.keras.layers.Dense(10,activation='softmax')
])
model_FNN.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
model_FNN.fit(x_train_SupervisedLearning,y_train_SupervisedLearning,epochs=10)

test_loss, test_accuracy = model_classFNN.evaluate(x_test,y_test,verbose=0)

print("Test Accuracy FNN:", test_accuracy)
```

```python
model_CNN = tf.keras.models.Sequential([
    tf.keras.Input(shape=(28,28,1)),
    tf.keras.layers.Conv2D(kernel_size=3,filters=32,activation='relu',padding='same'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(kernel_size=3,filters=64,activation='relu',padding='same'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(kernel_size=3,filters=128,activation='relu',padding='same'),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(100,activation='relu'),
    tf.keras.layers.Dense(100,activation='relu'),
    tf.keras.layers.Dense(100,activation='relu'),
    tf.keras.layers.Dense(10,activation='softmax'),
])

model_CNN.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
history = model_CNN.fit(x_train_SupervisedLearning.reshape(-1,28,28,1),y_train_SupervisedLearning,epochs=10)

test_loss, test_accuracy = model_CNN.evaluate(x_test.reshape(-1,28,28,1),y_test)

print("Test Accuracy CNN :", test_accuracy)
```

```python
inputs_encoder = tf.keras.Input(shape=(784,))
x = tf.keras.layers.Dense(256,activation='relu')(inputs_encoder)
x = tf.keras.layers.Dense(128,activation='relu')(x)
outputs_encoder = tf.keras.layers.Dense(32,activation='relu')(x)
encoder = tf.keras.Model(inputs_encoder,outputs_encoder,name='Encoder')

inputs_decoder = tf.keras.Input(shape=(32,))
x = tf.keras.layers.Dense(128,activation='relu')(inputs_decoder)
x = tf.keras.layers.Dense(256,activation='relu')(x)
outputs_decoder = tf.keras.layers.Dense(784,activation='sigmoid')(x)
decoder = tf.keras.Model(inputs_decoder,outputs_decoder,name='decode')

inputs_autoencoder = tf.keras.Input(shape=(784,))
outputs_autoencoder = decoder(encoder(inputs_autoencoder))
Autoencoder = tf.keras.Model(inputs_autoencoder,outputs_autoencoder,name='autoencoder')
Autoencoder.compile(loss='mse',optimizer='adam')
Autoencoder.fit(x_train_UnsupervisedLearning,x_train_UnsupervisedLearning,epochs=10)

encoder.trainable=False

model_Unsupervised_Pretraining = tf.keras.models.Sequential([
    tf.keras.Input(shape=(784,)),
    encoder,
    tf.keras.layers.Dense(100,activation='relu'),
    tf.keras.layers.Dense(100,activation='relu'),
    tf.keras.layers.Dense(100,activation='relu'),
    tf.keras.layers.Dense(10,activation='softmax')
]) 

model_Unsupervised_Pretraining.compile(loss=tf.keras.losses.SparseCategoricalCrossentropy(),optimizer='adam',metrics=['accuracy'])
model_Unsupervised_Pretraining.fit(x_train_SupervisedLearning,y_train_SupervisedLearning,epochs=30)
accuracy = model_Unsupervised_Pretraining.evaluate(x_test, y_test,verbose=0)
print("Test Accuracy Unsupervised Pretraining :", accuracy[1])

encoder.trainable = True 

model_Unsupervised_Pretraining.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),loss="sparse_categorical_crossentropy",metrics=["accuracy"])
history_finetune = model_Unsupervised_Pretraining.fit(x_train_SupervisedLearning,y_train_SupervisedLearning,epochs=30)
test_loss, test_accuracy = (model_Unsupervised_Pretraining.evaluate(x_test,y_test,verbose=0))
print("Fine-tuned Test Accuracy:", test_accuracy)
```
